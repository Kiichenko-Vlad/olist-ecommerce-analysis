## Q1: Динаміка кількості замовлень по місяцях

### Запит від продакт-менеджера
> *"У нас скоро зустріч з інвесторами, і мені треба показати їм як зростала наша платформа. Можеш зробити розбивку кількості замовлень по місяцях за весь період даних? Хочу побачити динаміку — чи дійсно ми ростемо, чи є сезонність (Black Friday, перед-різдвяні піки), і як виглядала траєкторія з 2016 по 2018 рік."*

### Висновок
Платформа демонструє стабільне зростання з кінця 2016 до середини 2018 року — кількість замовлень виросла з **~300/міс до ~6,000–7,000/міс** (~20x за 2 роки). Чітко видно піковий місяць — **листопад 2017** (Black Friday в Бразилії), що підтверджує наявність сезонності і свідчить що платформа вже має достатній масштаб щоб реагувати на ринкові події.

### Стратегічний інсайт
Динаміка відображає класичну S-криву зростання молодого маркетплейсу: повільний старт (2016) — прискорення (2017) — стабілізація на вищому рівні (2018). Листопадовий спайк 2017 — перша підтверджена точка сезонності — означає що планування логістичних і маркетингових ресурсів на Q4 має враховувати мінімум **1.5x базового навантаження**.

### Рекомендації
1. Планувати логістичні ресурси з урахуванням пікового навантаження у листопаді — заздалегідь збільшувати пропускну здатність у перевізників і на складах.
2. Запустити спеціальні маркетингові кампанії до Black Friday — органічний попит є, його треба підсилити.
3. Розширити збір даних за 2019+ для підтвердження чи зростання продовжилось або вийшло на плато.

### Застереження
Дані за **вересень–грудень 2016** та **серпень–жовтень 2018** неповні через обрізку датасету. Перший і останній місяці показують штучно низькі значення і **не відображають реального спаду бізнесу** — це артефакт даних, а не бізнес-сигнал.

## Q2: Топ-10 категорій товарів за доходом

### Запит від продакт-менеджера
> *"У нас обговорення з командою закупівель — куди інвестувати ресурси і з якими категоріями працювати агресивніше. Можеш витягнути ТОП-10 категорій за доходом? По сумі продажів (price), без урахування доставки — мене цікавить саме виручка від товарів. Назви категорій — англійською."*

### Висновок
Топ-3 категорії — **health_beauty** (~R$1.23M), **watches_gifts** (~R$1.17M), **bed_bath_table** (~R$1.02M) — генерують значно більше виручки ніж решта топ-10. Важливий патерн: частка логістичних витрат суттєво варіюється між категоріями — від **~8% у `watches_gifts`** до **~20% у `bed_bath_table`**. Це означає що за однаковою виручкою стоїть кардинально різна маржинальність.

### Стратегічний інсайт
Рейтинг за валовою виручкою не є рейтингом за прибутком. `bed_bath_table` займає 3 місце за виручкою, але витрачає майже чверть на доставку. Натомість `watches_gifts` — 2 місце при лише 8–9% логістичних витрат. **Стратегічно `watches_gifts` є пріоритетом для масштабування**: при тому самому маркетинговому бюджеті він дасть вищий чистий дохід.

### Рекомендації
1. Пріоритизувати маркетингові інвестиції у "легкі" категорії (`health_beauty`, `watches_gifts`) — вищий дохід при нижчій логістичній собівартості.
2. Для "важких" категорій (`bed_bath_table`, `furniture_decor`) — провести переговори з логістичним партнером щодо тарифів або розглянути спеціалізованих перевізників.
3. Ввести метрику `freight_pct_of_revenue` як обов'язковий показник при аналізі категорій — щоб порівнювати не лише виручку, а й маржинальність.

### Застереження
Виручка розрахована як `SUM(price)` без урахування комісій Olist, витрат на повернення та операційних витрат продавців. Реальна маржинальність категорій може суттєво відрізнятися від представленого рейтингу.

## Q3: Середній час доставки по штатах

### Запит від продакт-менеджера
> *"До нас дійшли скарги від клієнтів з північних штатів що доставка занадто довга. Перевір по даних — який середній час доставки (від покупки до отримання) у розрізі штатів? Хочу побачити топ найгірших регіонів, щоб поговорити з логістичним партнером і, можливо, відкрити склад ближче до проблемних зон."*

### Висновок
Аналіз доставки по 27 штатах підтверджує гіпотезу команди: найбільші затримки — у північних штатах (**RR — 29.4 дні**, AP — 27.2, AM — 26.4) проти **8.8 днів у SP** (різниця майже у 4 рази). Розбивка на три етапи локалізувала проблему: час оформлення продавцем стабільний (~3–4 дні) у всіх 27 штатах, тоді як **транзит перевізника варіюється від 5.6 (SP) до 25.6 днів (RR)** — різниця у 4.5 рази.

Додатковий інсайт: у північних штатах Olist виставляє реалістичні очікувані дати (`late_pct` помірний — RR: 12%). Натомість **AL (Алагоас) — критична зона**: при середньому часі 24.5 днів частка прострочених замовлень сягає **24%** — Olist обіцяє клієнтам швидше ніж може виконати.

### Стратегічний інсайт
Проблема **повністю на стороні логістичного партнера**, а не продавців. Розмова має бути з перевізником, а не з командою продавців (-> Q5). `late_pct` виявився кращим індикатором клієнтського невдоволення ніж абсолютний час: AL з 24% запізнень є більш критичним кейсом ніж RR з 12%, хоча RR фізично повільніший.

### Рекомендації
1. Пріоритизувати переговори з логістичним партнером щодо північного напрямку — розглянути окремий контракт або відкриття фулфілмент-хабу у Манаусі (AM) (-> Q9).
2. Переглянути алгоритм розрахунку `estimated_delivery_at` у штатах північного сходу (AL, MA, SE) — зменшити розрив між обіцянками і реальністю.
3. Не інвестувати у програми покращення seller-fulfillment — дані чітко показують що проблема не там.

### Застереження
Для штатів RR/AP/AC обсяг даних малий (41–80 замовлень) — результати слід підтверджувати на більшій вибірці. Аналіз проведено тільки для доставлених замовлень з наявною датою передачі перевізнику.

## Q4: Розподіл оцінок відгуків (1–5)

### Запит від продакт-менеджера
> *"Мені треба швидко зрозуміти — наскільки наші клієнти задоволені сервісом загалом? Зроби розподіл оцінок з відгуків у відсотках. Хочу зрозуміти яка частка негативу (1–2 зірки) у нас на платформі."*

### Висновок
Аналіз 98,410 відгуків показав загалом позитивну картину: **77.14% клієнтів ставлять 4–5 зірок**. Однак частка негативних відгуків (1–2 зірки) сягає **14.62%** — кожен 7-й клієнт незадоволений, що вище індустріального бенчмарку (~10%).

Ключова знахідка — чітка **U-форма розподілу**: оцінок "1" у **3.6 раза більше** ніж "2" (11,282 проти 3,114). Клієнти Olist поляризовані: ті, хто залишають негативний відгук, ставлять одразу мінімальну оцінку — це сигнал про критичні збої (втрачені посилки, серйозні проблеми з якістю), а не помірну незадоволеність.

### Стратегічний інсайт
U-форма означає що "помірного покращення" не існує — клієнт або отримав бездоганний досвід (5 зірок) або зіткнувся з критичним збоєм (1 зірка). Стратегічно: **точкова робота з усунення критичних збоїв (score=1) дасть більший ефект на NPS** ніж дифузне "загальне покращення сервісу". Двійки — рідкісний і менш інформативний сигнал (-> Q5, де цей патерн підтвердився на рівні індивідуальних продавців).

### Рекомендації
1. Запровадити окремий процес ескалації для одиничних оцінок — це найбільш концентрована категорія проблем.
2. Проаналізувати текстові коментарі при `score=1` щоб виявити топ-3 корінних причини.
3. Зменшення кількості одиничних оцінок — найшвидший важіль для покращення NPS платформи.

### Застереження
Дані містять selection bias — клієнти з негативним досвідом схильні залишати відгуки частіше. Тому реальний відсоток незадоволених серед усіх клієнтів (включно з тими хто не залишив відгуку) імовірно нижчий за виміряні 14.62%.

## Q5: Топ-10 продавців за кількістю замовлень та середня оцінка

### Запит від продакт-менеджера
> *"Команда роботи з продавцями попросила список найактивніших продавців — тих хто продає найбільше. Витягни ТОП-10 за кількістю замовлень. Але важливий нюанс: покажи одночасно їх середню оцінку від клієнтів. Підозрюю що хтось женеться за обсягом і жертвує якістю — хочу це побачити в одній таблиці."*

### Висновок
Топ-10 найактивніших продавців Olist обробили від 1,076 до 1,847 замовлень. **Усі 10 — у штаті SP**, що відображає сильну економічну концентрацію і потенційний географічний ризик (-> Q9).

Кореляція "обсяг / якість" виявилася **слабкою**: AVG-оцінка коливається від 3.81 до 4.23, `critical_pct` (частка одиничних оцінок) — від 7.74% до 14.73% (майже у 2 рази). Найбільший за обсягом продавець (#1, 1,847 замовлень) має лише посередні показники якості (AVG 3.92, critical 12.97%). Продавець #7 (1,159 замовлень, sao paulo) — позитивний outlier: AVG 4.23, лише 7.74% критичних відгуків. У тривожній зоні — #2 (ibitinga, 14.73%) і #4 (sao jose do rio preto, 14.54%): кожен 7-й клієнт ставить 1 зірку.

Поведінкова знахідка: `critical/negative` ratio для всіх продавців = 0.71–0.80, тобто **3 з 4 негативних відгуків — це одиниці** (підтверджує U-форму з -> Q4).

### Стратегічний інсайт
AVG-оцінка як єдина метрика якості є недостатньою і **маскує системні проблеми**. Продавці #2 і #4 мають AVG ~3.8–4.0 (виглядає прийнятно), але `critical_pct` вдвічі вищий ніж у #7. Без `critical_pct` ці продавці виглядають "нормальними" — з нею видно що кожен 7-й їхній клієнт обурений.

### Рекомендації
1. Запровадити моніторинг `critical_pct` на додачу до AVG як ранній індикатор погіршення сервісу.
2. Для продавців у тривожній зоні (#2, #4) — провести аудит причин критичних відгуків (гіпотеза: специфіка важких категорій з довгою доставкою -> Q8).
3. Запустити програму "premium seller" для #7 як публічний приклад і механізм мотивації.
4. Диверсифікувати географію топ-продавців: 100% у SP — критичний ризик концентрації (-> Q9).

### Застереження
У multi-seller orders (один клієнт купив у двох продавців) відгук один на все замовлення — неможливо визначити якому продавцю він адресований. Частка таких замовлень в Olist невелика, але може дещо спотворювати індивідуальні метрики якості.

## Q6: Частка повторних клієнтів (Retention)

### Запит від продакт-менеджера
> *"Для квартального звіту мені потрібна метрика retention. Яка частка наших клієнтів зробила більше одного замовлення? Це буде один з ключових KPI у презентації для CEO."*

### Висновок
Аналіз 96,096 унікальних клієнтів виявив **критично низький рівень утримання — лише 3.12%** клієнтів повернулися за повторною покупкою. Геометричний обвал по мірі вимог лояльності:

| Сегмент | Кількість | % від усіх |
|---|---|---|
| Одноразові | 93,099 | 96.88% |
| Повторні (2+) | 2,997 | 3.12% |
| Часті (3+) | 252 | 0.26% |
| Лояльні (4+) | 49 | 0.05% |
| Power users (5+) | 19 | 0.02% |

Середня кількість замовлень на клієнта — **1.03**. Типовий клієнт Olist приходить рівно один раз.

### Стратегічний інсайт
При retention 3% і зростаючому CAC Olist має **структурну вразливість**: кожна гривня маркетингового бюджету відбивається рівно одним замовленням — немає LTV для амортизації витрат. Для порівняння: клієнт з 2 покупками генерує вдвічі більше виручки при тому самому CAC. Збільшення retention з 3% до 6% (подвоєння) дало б платформі орієнтовно **~3,000 додаткових замовлень на рік** без жодних витрат на залучення. **Retention — найвищий ROI-важіль з усіх можливих у поточній ситуації.**

### Рекомендації
1. Запровадити програму лояльності (бонуси, кешбек) для стимуляції другої покупки — стрибок з 1 до 2 замовлень є найдорожчим і найважливішим.
2. Email-ретаргетинг через 30/60/90 днів після першої покупки з релевантними категоріями.
3. Дослідити 19 power users — опитування щоб зрозуміти що утримало їх на платформі і масштабувати ці фактори.
4. Cross-sell стратегія: одна покупка часто закриває потребу в категорії на роки (матрац, меблі) — пропонувати суміжні категорії одразу після покупки.

### Застереження
Через анонімізацію `customer_unique_id` фактичний retention може бути дещо вищим — один реальний клієнт теоретично міг отримати кілька ідентифікаторів. Обрізаний датасет (2016–2018) також робить "одноразовими" тих клієнтів, які не встигли повернутися до кінця досліджуваного періоду.

## Q7: Розподіл способів оплати та середній чек

### Запит від продакт-менеджера
> *"Фінансовий відділ просить розібратися з платіжними методами. Покажи для кожного типу: кількість транзакцій, частку від загального обсягу і середній чек. Цікаво — ті хто платять кредиткою купують на більшу суму ніж ті хто через boleto? Це допоможе нам вирішити які методи просувати у чекауті."*

### Висновок
**Credit card домінує — 75.24% транзакцій** (76,505), що пояснюється унікальною бразильською практикою *parcelamento* (розстрочки до 12 місяців без процентів для клієнта). **Boleto займає 19.46%** — критичний канал для ~20% клієнтів без credit card. **Debit card — лише 1.50%**: без розстрочки клієнти обирають credit навіть маючи кошти на дебеті. **Voucher (3.80%)** — не самостійний метод, а інструмент часткової оплати (промокоди, повернення), що пояснює найнижчий середній чек (R$98.15).

Ключовий патерн: середні чеки credit_card (R$163.94) і boleto (R$145.03) майже однакові — різниця лише 13%. Тип оплати **не корелює з сумою замовлення** — клієнти мають сталі платіжні переваги незалежно від суми. Це різні сегменти аудиторії, а не різні поведінкові режими.

### Стратегічний інсайт
Оскільки avg_order_value credit_card і boleto майже однакові, "просування" одного методу за рахунок іншого **не збільшить середній чек**. Boleto-клієнти — окремий сегмент з нижчим рівнем фінансової інклюзії: відключення цього методу означає втрату ~20% аудиторії **без жодної компенсації** з боку credit_card користувачів.

### Рекомендації
1. Зберегти всі чотири методи активними — кожен обслуговує свій сегмент. Відключення boleto особливо ризиковане.
2. За замовчуванням пропонувати credit_card з розстрочкою у checkout — найпопулярніший вибір (75%).
3. Перевірити кореляцію вибору boleto з географією — можливе нижче проникнення кредитних карток у певних штатах (-> Q9), що може дати інсайт для регіональних кампаній.

### Застереження
Аналіз проведено на рівні `(order_id, payment_type)` — одне замовлення з двома методами оплати зараховується двічі. Загальна кількість записів (~102k) перевищує кількість замовлень (~99k). 3 записи з `payment_type = 'not_defined'` (value=0) є технічною аномалією і не впливають на висновки.

## Q8: Чи впливає час доставки на оцінку клієнта?

### Запит від продакт-менеджера
> *"У мене є гіпотеза що повільна доставка вбиває наш рейтинг — клієнти гніваються коли довго чекають і ставлять низькі оцінки. Перевір: розбий замовлення на бакети за часом доставки і для кожного покажи середню оцінку. Якщо підтвердиться — виставимо SLA для логістики на основі реальних даних."*

### Висновок
Аналіз 96,707 доставлених замовлень **підтвердив гіпотезу**: час доставки прямо і монотонно впливає на оцінку клієнта.

| Бакет | Замовлень | Частка | Avg score |
|---|---|---|---|
| 1–3 дні | 8,582 | 8.9% | **4.46** |
| 4–7 днів | 25,124 | 26.0% | 4.39 |
| 8–14 днів | 36,517 | 37.8% | 4.29 |
| **15+ днів** | **26,484** | **27.4%** | **3.65** |

Залежність **нелінійна**: між бакетами 1–3, 4–7, 8–14 падіння поступове (~0.07–0.10 за крок), але після 15 днів — **різкий обвал на 0.64 пункти** (у 6 разів більше). Це класичний **пороговий ефект**. Масштаб: кожне четверте замовлення (27.4%) потрапляє у критичний бакет 15+.

### Стратегічний інсайт
Порогова точка **15 днів** — це конкретна операційна ціль. Стратегія не "доставляти якомога швидше" (дорого і нереалістично для периферії), а **"утримувати доставку до 15 днів"** — досяжний і вимірюваний SLA. В поєднанні з Q3 (причина — перевізник, не продавець) і Q9 (північ має вищий avg_order_value) отримуємо повний інвестиційний кейс для логістичної реструктуризації.

### Рекомендації
1. Встановити SLA з логістичним партнером: максимум **14 днів** для всіх маршрутів.
2. Розробити окремий контракт для північних штатів де поточний середній час систематично перевищує 15-денний поріг (-> Q3).
3. Ввести автоматичні компенсації (промокод, знижка) при перевищенні SLA — частково амортизує незадоволеність і підвищує шанс повторної покупки (-> Q6).

### Застереження
`avg_score` розраховується тільки по замовленнях де клієнт залишив відгук. Клієнти з довгою доставкою могли відмовитись від відгуку взагалі (disengagement) — тому реальна незадоволеність у бакеті 15+ може бути **ще вищою** ніж показує оцінка 3.65.

## Q9: Географія доходів — топ штатів за GMV

### Запит від продакт-менеджера
> *"Хочемо зрозуміти географічну структуру нашого бізнесу: які штати генерують найбільшу виручку? І якщо можна — показати також середній чек по штатах, щоб розуміти не тільки обсяг але і цінність клієнта в кожному регіоні."*

### Висновок
**SP (Сан-Паулу) генерує ~40% всієї виручки** платформи (R$5.16M з ~R$12.8M загальних), топ-3 штати (SP + RJ + MG) концентрують **~66% GMV** — решта 24 штатів ділять третину ринку. Це підтверджує висновки Q3 (найкраща доставка у SP/MG/PR) та Q5 (всі топ-10 продавців з SP): **Olist фактично є бізнесом одного регіону**.

Несподіваний патерн — **інверсна залежність між GMV і середнім чеком**: штати з найменшим обсягом мають найвищий avg_order_value (PB — R$216.34, AP — R$198.15, AC — R$197.32 проти R$125.57 у SP, різниця **+73%**). Периферійні клієнти замовляють рідше, але дорожче.

### Стратегічний інсайт
Штати з найгіршою доставкою (RR — 29.4 дні, AP — 27.2, AM — 26.4) знаходяться в кінці рейтингу GMV, але їхній **avg_order_value вищий ніж у SP** (171–198 R$ проти 125 R$). Проблема не у відсутності попиту чи платоспроможності — проблема у **логістичному бар'єрі**. Покращення доставки на північ (-> Q3, Q8) — це не соціальна відповідальність, а **розблокування ринку з вищою цінністю транзакції**.

### Рекомендації
1. **Диверсифікувати географію продавців**: 100% топ-10 у SP при 40% GMV з одного штату — критичний ризик концентрації (будь-який локальний збій б'є одночасно по постачанні і попиті).
2. **Пріоритет логістичних інвестицій — північ**: відкриття хабу в Манаусі (AM) або Белемі (PA) розблоковує ринок з вищим avg_order_value ніж основний регіон (-> Q3).
3. **Маркетингові кампанії для NE-штатів** (CE, PE, BA, PB): вже генерують помітний GMV (R$226–507K), потенціал увійти до топ-5 при покращенні логістики.
4. Встановити KPI **"частка SP у GMV < 35%"** як індикатор здорової географічної диверсифікації платформи.

### Застереження
GMV розраховано як `SUM(price)` без урахування `freight_value`. Включення доставки збільшило б відносну вагу північних штатів (вища вартість логістики) і дещо змінило б ранжування — загальних висновків не змінює.

## 7.3 Обмеження аналізу

При інтерпретації висновків необхідно враховувати такі обмеження:

**1. Актуальність даних**  
Датасет охоплює період вересень 2016 — жовтень 2018. Бразильський e-commerce ринок суттєво змінився після 2018 року (зростання конкурентів, COVID-19, розвиток логістичної інфраструктури). Висновки відображають стан платформи на той момент і потребують перевірки на актуальних даних перед прийняттям стратегічних рішень.

**2. Відсутність фінансових даних**  
Аналіз оперує виручкою (`price`) без урахування комісій Olist, логістичних субсидій, витрат на повернення, операційних витрат продавців і CAC (вартості залучення клієнта). Рейтинги категорій і регіонів за "виручкою" можуть кардинально відрізнятись від рейтингів за реальною маржинальністю.

**3. Анонімізація клієнтів**  
В датасеті `customer_id` (PK замовлення) і `customer_unique_id` (реальний клієнт) є різними полями. Метрика retention 3.12% розрахована на основі `customer_unique_id` — проте анонімізація не виключає, що один реальний клієнт у поодиноких випадках отримав кілька `customer_unique_id`. Реальний retention може бути незначно вищим.

**4. Selection bias у відгуках**  
Відгук залишає лише частина клієнтів. Клієнти з негативним досвідом схильні відгукуватись частіше — тому виміряна частка незадоволених (14.62%) є верхньою межею, а реальне значення серед усіх клієнтів, ймовірно, нижче. З іншого боку, клієнти з дуже довгою доставкою могли взагалі не залишити відгук (disengagement), що може занижувати реальну незадоволеність у бакеті 15+.

**5. Малий обсяг по окремих штатах**  
Для штатів RR (41 замовлення), AP (77) та AC (80) розраховані середні значення (час доставки, avg_order_value) є статистично ненадійними через малу вибірку. Висновки по цих штатах слід розглядати як індикативні, а не остаточні.

**6. Відсутність даних про маркетингові канали**  
Аналіз не включає інформацію про канали залучення клієнтів (органіка, paid, referral). Без цих даних неможливо точно розрахувати CAC і LTV по сегментах, що обмежує глибину рекомендацій щодо retention та програм лояльності.

**7. Відгук на замовлення, а не на продавця**  
В Olist один відгук прив'язаний до всього замовлення, навіть якщо воно містить товари від кількох продавців. У Q5 метрики якості продавців (`avg_score`, `critical_pct`) можуть бути частково спотворені у випадку multi-seller orders.

## Головний системний висновок

Три незалежних аналізи — Q3 (час доставки по штатах), Q8 (вплив доставки на оцінку) і Q9 (географія GMV) — вказують в одну точку: **логістичний партнер на північ та північний схід Бразилії є найкритичнішим вузьким місцем платформи**.

Вирішення цієї проблеми через встановлення SLA 14 днів та відкриття фулфілмент-хабу на північ одночасно:
- покращить `avg_score` по платформі (Q8 — переведення замовлень з бакету 15+ у прийнятні бакети),
- розблокує ринок з вищим `avg_order_value` (Q9 — північ: R$171–198 vs SP: R$126),
- знизить `late_pct` у найпроблемніших штатах (Q3 — RR, AP, AL),
- та потенційно підвищить retention (Q6) через покращення клієнтського досвіду після першої покупки.

Це єдина рекомендація, яка одночасно впливає на всі чотири ключових метрики платформи: `avg_score`, `GMV`, `late_pct` та `retention`.
